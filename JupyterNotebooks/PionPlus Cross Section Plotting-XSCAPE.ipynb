```python
def align_and_ratio(x1, y1, y1_err, x2, y2, y2_err):
    # Find common x-range
    x_min = max(min(x1), min(x2))
    x_max = min(max(x1), max(x2))
    
    # Use x1 as reference
    # x_ref = x1[(x1 >= x_min) & (x1 <= x_max)]
    # y1_ref = y1[(x1 >= x_min) & (x1 <= x_max)]
    x_ref = np.linspace(x_min, x_max, num=max(len(x1), len(x2)))
    y1_interp = np.interp(x_ref, x1, y1)
    
    # Interpolate y2 to match x_ref
    y2_interp = np.interp(x_ref, x2, y2)
    y1_err_interp = np.interp(x_ref, x1, y1_err)
    y2_err_interp = np.interp(x_ref, x2, y2_err)

    # Calculate ratio
    # ratio = y1_ref / y2_interp
    ratio = y1_interp / y2_interp
    ratio_err = ratio * np.sqrt((y1_err_interp/y1_interp)**2 + (y2_err_interp/y2_interp)**2)

    return x_ref, ratio, ratio_err

# Usage
# x1 = existing_pionall_pT
# y1 = existing_pionall_cross_sec
# y1_err = np.sqrt(existing_pionall_stat_err**2+existing_pionall_syst_err**2)
x1 = existing_pionplus_pT
y1 = existing_piontotal_cross_sec
y1_err = existing_piontotal_cross_sec_err
x2 = pionplusminus_pythia_15_30_cut_pT
y2 = pionplusminus_pythia_5_30_comb_cut_cross_sec
y2_err = pionplusminus_pythia_5_30_comb_cut_cross_sec_err

x_aligned, ratio, ratio_err = align_and_ratio(x1, y1, y1_err, x2, y2, y2_err)
# plt.plot(x_aligned, ratio, color='black', label='PHENIX $\pi^{0}$ $|\eta|<0.35$')
plt.errorbar(x_aligned, 2*ratio, yerr=ratio_err, fmt='o', color='black', capsize=5, label='PHENIX $(\pi^{+}+\pi^{-})/2$ $|\eta|<0.35$')

xscape_pythia_ratio_err = (pionplusminus_xscape_5_30_comb_cut_cross_sec/pionplusminus_pythia_5_30_comb_cut_cross_sec) * np.sqrt((pionplusminus_xscape_5_30_comb_cut_cross_sec_err/pionplusminus_xscape_5_30_comb_cut_cross_sec)**2 + (pionplusminus_pythia_5_30_comb_cut_cross_sec_err/pionplusminus_pythia_5_30_comb_cut_cross_sec)**2)
pythia_pythia_ratio_err = (pionplusminus_pythia_5_30_comb_cut_cross_sec/pionplusminus_pythia_5_30_comb_cut_cross_sec) * np.sqrt((pionplusminus_pythia_5_30_comb_cut_cross_sec_err/pionplusminus_pythia_5_30_comb_cut_cross_sec)**2 + (pionplusminus_pythia_5_30_comb_cut_cross_sec_err/pionplusminus_pythia_5_30_comb_cut_cross_sec)**2)

plt.errorbar(pionplusminus_xscape_15_30_cut_pT, (pionplusminus_xscape_5_30_comb_cut_cross_sec)/pionplusminus_pythia_5_30_comb_cut_cross_sec, yerr=xscape_pythia_ratio_err, fmt='-', color='red', capsize=5, label='XSCAPE $\hat{p_{T}}$ 5-30 comb $(\pi^{+}+\pi^{-})/2$')
plt.errorbar(pionplusminus_pythia_15_30_cut_pT, (pionplusminus_pythia_5_30_comb_cut_cross_sec)/pionplusminus_pythia_5_30_comb_cut_cross_sec, yerr=pythia_pythia_ratio_err, fmt='-.', color='blue', capsize=5, label='PYTHIA $\hat{p_{T}}$ 5-30 comb $(\pi^{+}+\pi^{-})/2$')
# plt.errorbar(existing_pionminus_pT, (existing_piontotal_cross_sec)/pionplusminus_pythia_5_30_comb_cut_cross_sec, yerr=existing_piontotal_cross_sec_err, fmt='.', color='black', capsize=5, label='PHENIX $(\pi^{+}+\pi^{-})/2$ $|\eta|<0.35$')

plt.ylim(-1,2.5)
plt.xlim(4.9,12)
# plt.title('Ratio to PYTHIA (Cross Section of $\pi^{+}$)')
plt.xlabel('$p_T$ (GeV)',fontsize=12)
plt.ylabel(r'Ratio to PYTHIA 8.309 [$E \frac{d^{3}\sigma}{dp^{3}}$]',fontsize=12)
# plt.title('Direct Photon Production at RHIC $\sqrt{s}=200GeV$ compared to data')
plt.legend(fontsize=9)
# plt.text(10.5, 1.7, 'pp, $\sqrt{s_{NN}} = 200$ GeV', fontsize=12)
# plt.text(10.5, 1.5, '$|\eta| < 0.35$, $(\pi^{+}+\pi^{-})/2$', fontsize=12)

plt.show()
```

```python
def align_and_ratio(x1, y1, y1_err, x2, y2, y2_err):
    # Find common x-range
    x_min = max(min(x1), min(x2))
    x_max = min(max(x1), max(x2))
    
    # Use x1 as reference
    # x_ref = x1[(x1 >= x_min) & (x1 <= x_max)]
    # y1_ref = y1[(x1 >= x_min) & (x1 <= x_max)]
    x_ref = np.linspace(x_min, x_max, num=max(len(x1), len(x2)))
    y1_interp = np.interp(x_ref, x1, y1)
    
    # Interpolate y2 to match x_ref
    y2_interp = np.interp(x_ref, x2, y2)
    y1_err_interp = np.interp(x_ref, x1, y1_err)
    y2_err_interp = np.interp(x_ref, x2, y2_err)

    # Calculate ratio
    # ratio = y1_ref / y2_interp
    ratio = y1_interp / y2_interp
    ratio_err = ratio * np.sqrt((y1_err_interp/y1_interp)**2 + (y2_err_interp/y2_interp)**2)

    return x_ref, ratio, ratio_err

# Usage
x1 = existing_pionall_pT
y1 = existing_pionall_cross_sec
y1_err = np.sqrt(existing_pionall_stat_err**2+existing_pionall_syst_err**2)
x2 = pionplusminus_pythia_15_30_pT
y2 = pionplusminus_pythia_5_30_comb_cross_sec
y2_err = pionplusminus_pythia_5_30_comb_cross_sec_err

x_aligned, ratio, ratio_err = align_and_ratio(x1, y1, y1_err, x2, y2, y2_err)
# plt.plot(x_aligned, ratio, color='black', label='PHENIX $\pi^{0}$ $|\eta|<0.35$')
plt.errorbar(x_aligned, ratio, yerr=ratio_err, fmt='o', color='black', capsize=5, label='PHENIX $\pi^{0}$ $|\eta|<0.35$')

xscape_yasuki_pythia_ratio_err = (pionplusminus_xscape_5_30_yasuki_comb_cross_sec/pionplusminus_pythia_5_30_comb_cross_sec) * np.sqrt((pionplusminus_xscape_5_30_yasuki_comb_cross_sec_err/pionplusminus_xscape_5_30_yasuki_comb_cross_sec)**2 + (pionplusminus_pythia_5_30_comb_cross_sec_err/pionplusminus_pythia_5_30_comb_cross_sec)**2)
xscape_pythia_ratio_err = (pionplusminus_xscape_5_30_comb_cross_sec/pionplusminus_pythia_5_30_comb_cross_sec) * np.sqrt((pionplusminus_xscape_5_30_comb_cross_sec_err/pionplusminus_xscape_5_30_comb_cross_sec)**2 + (pionplusminus_pythia_5_30_comb_cross_sec_err/pionplusminus_pythia_5_30_comb_cross_sec)**2)
pythia_pythia_ratio_err = (pionplusminus_pythia_5_30_comb_cross_sec/pionplusminus_pythia_5_30_comb_cross_sec) * np.sqrt((pionplusminus_pythia_5_30_comb_cross_sec_err/pionplusminus_pythia_5_30_comb_cross_sec)**2 + (pionplusminus_pythia_5_30_comb_cross_sec_err/pionplusminus_pythia_5_30_comb_cross_sec)**2)

plt.errorbar(pionplusminus_xscape_15_30_pT, (pionplusminus_xscape_5_30_comb_cross_sec/2)/(pionplusminus_pythia_5_30_comb_cross_sec), yerr=xscape_yasuki_pythia_ratio_err, fmt='-', color='red', capsize=5, label='XSCAPE $\hat{p_{T}}$ 5-30 comb $(\pi^{+}+\pi^{-})/2$')
plt.errorbar(pionplusminus_xscape_15_30_yasuki_pT, (pionplusminus_xscape_5_30_yasuki_comb_cross_sec/2)/(pionplusminus_pythia_5_30_comb_cross_sec), yerr=xscape_pythia_ratio_err, fmt='-', color='purple', capsize=5, label='XSCAPE $\hat{p_{T}}$ 5-30 comb Yasuki sample $(\pi^{+}+\pi^{-})/2$')
plt.errorbar(pionplusminus_pythia_15_30_pT, (pionplusminus_pythia_5_30_comb_cross_sec/2)/(pionplusminus_pythia_5_30_comb_cross_sec), yerr=pythia_pythia_ratio_err, fmt='-.', color='blue', capsize=5, label='PYTHIA $\hat{p_{T}}$ 5-30 comb $(\pi^{+}+\pi^{-})/2$')

plt.ylim(-1,2.5)
plt.xlim(3,21)
# plt.title('Ratio to PYTHIA (Cross Section of $\pi^{+}$)')
plt.xlabel('$p_T$ (GeV)',fontsize=12)
plt.ylabel(r'Ratio to PYTHIA 8.309 [$E \frac{d^{3}\sigma}{dp^{3}}$]',fontsize=12)
# plt.title('Direct Photon Production at RHIC $\sqrt{s}=200GeV$ compared to data')
plt.legend(fontsize=9)
# plt.text(10.5, 1.7, 'pp, $\sqrt{s_{NN}} = 200$ GeV', fontsize=12)
# plt.text(10.5, 1.5, '$|\eta| < 0.35$, $(\pi^{+}+\pi^{-})/2$', fontsize=12)

plt.show()
```
